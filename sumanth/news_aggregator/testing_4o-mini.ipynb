```python
# from openai import OpenAI
import google.generativeai as genai
from dotenv import load_dotenv
import os
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
```

```python
# api_key = os.getenv("openai_api_key")
# client = OpenAI(api_key=api_key)

api_key = os.getenv("google_api_key")
genai.configure(api_key=api_key)
model = genai.GenerativeModel("gemini-1.5-pro")
# client = OpenAI(api_key=api_key)
```

```python
system_prompt = """
You are a news summarizer that fetches recent, real-world news only from reliable, verified sources. Do not generate or invent news.

Your job:
1. Only summarize existing news content.
2. Prioritize:
   - Recency (prefer last 48h)
   - Relevance to user's interests (if provided)
   - Source credibility
   - Non-duplicate, non-clickbait articles

Behavior:
- Maintain factual accuracy and neutrality.
- Discard opinion pieces, unverified claims, or low-trust sources.
- Do not hallucinate or add unconfirmed content.

Personalization (when possible):
- Use user interests, location, or language to rank relevance.

Output format for each story:
- Headline
- Source Link(must work, the user able to read full article)
- Time (e.g., "3h ago")
- Three-line summary

Only include informative, fact-based news. Keep output brief and structured.
"""
```

```python
user_prompt = """
{
  "intent": "fetch_news",
  "user_profile": {
    "language": "en",
    "preferred_category": "Technology",
    "audience_type": "creator"
  },
  "content_preferences": {
    "depth": "brief",
    "sentiment": "informative",
    "published_within": "48h",
    "min_trending_score": 70,
    "source_constraints": {
      "verified_sources": true
    }
  },
  "delivery_preferences": {
    "include_links": true,
    "formatting_style": "bullet"
  }
}

Based on this, give me the top 5 tech news summaries in bullet format from verified sources published within 48 hours. Keep it informative and brief.
"""
```

```python
# messages = [{"role" : "system", "content" : system_prompt}] +  [{"role" : "user", "content" : user_prompt}]
full_prompt = system_prompt.strip() + "\n\n" + user_prompt.strip()
```

```python
# response = client.chat.completions.create(
#     model = "gpt-4o-mini",
#     messages = messages
# )

response = model.generate_content(system_prompt)
```
